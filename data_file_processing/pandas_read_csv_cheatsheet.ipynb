# Pandas `read_csv` Cheat Sheet
Most commonly used parameters with examples.

## Basic Usage

```
import pandas as pd
# df = pd.read_csv('data.csv')
```

## Header and Names

```
pd.read_csv('data.csv', header=None)
pd.read_csv('data.csv', names=['id','name','age'])
```

## Selecting Columns

```
pd.read_csv('data.csv', usecols=['id','age'])
```

## Missing Values

```
pd.read_csv('data.csv', na_values=['NA','null'])
```

## Data Types

```
pd.read_csv('data.csv', dtype={'id': int, 'price': float})
```

## Date Parsing

```
pd.read_csv('data.csv', parse_dates=['order_date'])
```

## Large File Handling

```
for chunk in pd.read_csv('bigfile.csv', chunksize=10000):
    print(chunk.shape)
```

## Delimiter

```
pd.read_csv('data.tsv', sep='\t')
```

## Encoding

```
pd.read_csv('data.csv', encoding='latin1')
```

## Bad Lines

```
pd.read_csv('data.csv', on_bad_lines='skip')
```
