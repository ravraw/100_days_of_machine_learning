## Reading CSV from External URL

You can also read CSV files directly from URLs using pandas. This is useful when working with datasets hosted online.

```python
# Example: Reading CSV from a URL
# Using a sample dataset from GitHub (Iris dataset)
url = 'https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv'

# Read CSV directly from URL
df_url = pd.read_csv(url)

print("Dataset shape:", df_url.shape)
print("\nFirst 5 rows:")
print(df_url.head())
print("\nDataset info:")
print(df_url.info())
```

```python
# Another example: Reading CSV with additional parameters from URL
# Using a different dataset (Tips dataset)
tips_url = 'https://raw.githubusercontent.com/mwaskom/seaborn-data/master/tips.csv'

# Read with specific parameters
df_tips = pd.read_csv(tips_url, 
                     sep=',',           # separator
                     encoding='utf-8',  # encoding
                     na_values=['NA', 'null', ''])  # custom NA values

print("Tips dataset:")
print("Shape:", df_tips.shape)
print("\nColumn names:")
print(df_tips.columns.tolist())
print("\nData types:")
print(df_tips.dtypes)
print("\nSample data:")
print(df_tips.head(3))
```
