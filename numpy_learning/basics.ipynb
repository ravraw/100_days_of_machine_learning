```python
#Creating numpy/n-d arrays
import numpy as np
```

```python
arr1=np.array([1, 2, 3, 4, 5])
print("1D Array:")
print(arr1)
print("Shape:", arr1.shape)
```

```python
arr2=np.array([[1, 2, 3], [4, 5, 6]])
print("\n2D Array:")
print(arr2)
print("Shape:", arr2.shape)
```

```python
arr3=np.zeros((3, 4))
print("\n2D Array of Zeros:")
print(arr3)
print("Shape:", arr3.shape)
```

```python
arr4=np.ones((2, 3))
print("\n2D Array of Ones:")
print(arr4)
print("Shape:", arr4.shape)
```

```python
aar5=np.identity(4)
print("\nIdentity Matrix:")
print(aar5)
print("Shape:", aar5.shape)
```

```python
arr6=np.random.rand(2, 3)
print("\n2D Array with Random Values:")
print(arr6)
print("Shape:", arr6.shape)
```

```python
arr7=np.arange(10,101,10)
print("\n1D Array with Range of Values:")
print(arr7)
print("Shape:", arr7.shape) 
```

```python
arr8=np.linspace(0, 1, 5)
print("\n1D Array with Linearly Spaced Values:")
print(arr8)
print("Shape:", arr8.shape) 
```

```python
arr9=np.copy(arr8)
print("\nCopied Array:")
print(arr9)
print("Shape:", arr9.shape) 
```
